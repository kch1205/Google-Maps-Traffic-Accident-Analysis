```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("cleaned_accidents.csv")
```

```python
df.columns.tolist()
```

```python
df['date'] = pd.to_datetime(df['date'])
```

```python
df['year'] = df['date'].dt.year
```

```python
df['month'] = df['date'].dt.month
```

```python
df['month_name'] = df['date'].dt.month_name()
```

```python
df['day'] = df['date'].dt.day
```

```python
df['weekday'] = df['date'].dt.day_name()
```

```python
df['quarter'] = df['date'].dt.quarter
```

```python
df['time'].head()
```

```python
df['time'].dtype
```

```python
df['time'] = pd.to_datetime(df['time'])
```

```python
df['hour'] = df['time'].dt.hour
```

```python
df[['time', 'hour']].head()
```

```python
def time_bucket(hour):
    if 5 <= hour < 12:
        return "Morning"
    elif 12 <= hour < 17:
        return "Afternoon"
    elif 17 <= hour < 21:
        return "Evening"
    else:
        return "Night"

df['time_bucket'] = df['hour'].apply(time_bucket)
```

```python
df[['hour', 'time_bucket']].head()
```

```python
df['is_weekend'] = df['weekday'].isin(
    ['Saturday', 'Sunday']
)
```

```python
df.head()
```

```python
df.to_csv(
    "feature_engineered_accidents.csv",
    index=False
)
```
